#### Task 2:Build Advanced Preprocessing Function


```python
def preprocess_text(text):
    if not isinstance(text, str) or text.strip() == "":
        return []

    # Lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # Remove emails
    text = re.sub(r"\S+@\S+", "", text)

    # Remove numbers
    text = re.sub(r"\d+", "", text)

    # Remove emojis & special characters
    text = re.sub(r"[^\w\s]", "", text)

    # Handle repeated characters 
    text = re.sub(r"(.)\1{2,}", r"\1\1", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # Tokenization
    tokens = text.split()

    # Remove short words 
    tokens = [word for word in tokens if len(word) > 2 or word in ["no", "not"]]

    return tokens
```

#### Task 3: Stress Testing


```python
test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

results = []

for sentence in test_sentences:
    tokens = preprocess_text(sentence)
    cleaned = " ".join(tokens)

    print("Original Text:", sentence)
    print("Cleaned Tokens:", tokens)
    print("Cleaned Sentence:", cleaned)
    print("-"*50)

    results.append(tokens)
```

#### Task 4: Token Analytics


```python
import numpy as np

for i, tokens in enumerate(results):
    if len(tokens) == 0:
        continue

    total = len(tokens)
    unique = len(set(tokens))
    avg_len = sum(len(word) for word in tokens) / total

    print(f"Sentence {i+1}")
    print("Total Tokens:", total)
    print("Unique Tokens:", unique)
    print("Avg Token Length:", round(avg_len, 2))
    print("-"*40)
```

#### Task 5: Frequency Analysis


```python
all_tokens = [word for sublist in results for word in sublist]

counter = Counter(all_tokens)

print("Top 10 words:", counter.most_common(10))

print("Least 5 words:", counter.most_common()[-5:])
```

#### Task 6: Build Full Pipeline


```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens = preprocess_text(text)
        clean_sentences.append(" ".join(tokens))
        all_tokens.extend(tokens)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }

output = full_pipeline(test_sentences)

print(output)
```
